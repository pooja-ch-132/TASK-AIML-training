**Python syntax, datatypes and variables**

Python syntax refers to rules used to write Python programs.

Rules:-

*   Indentation : Python uses indentation(spaces) to define blocks of code.
*  Comments : Single -line comments starts with "#", Multi line comment starts with """comment"""
*   Case sensitive : "John" and "john" are not same. Variables in Python store data values. Creating variables, Printing variables, multiple variables.

Python Datatypes:-
int 10, float 10.0, string "Hello", bool True, list [1,2,3], tuple (1,2,3), set{1,2,3}, dict {"name":"Pooja"}





```python
num = 10
print(type(num))
```

```python
#Initializing multiple variables
x=10
y=20
sum=x+y
print(sum)
```

```python
# Note: variables should not start with numerics. should not be predefined datatype & keywords like int, float.
x=y=z=100
print(x)
print(y)
print(z)
```

```python
#Float
percent = 78.8
print(percent)
```

```python
#Tuple
nums = (1,2,3,4,5,4,5)
print(nums)
```
