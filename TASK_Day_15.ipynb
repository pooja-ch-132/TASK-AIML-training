**Introduce Supervised Learning:**

Definition:

Supervised Learning is a type of Machine Learning where the model learns from labeled data and predicts outputs for new data.

| Input         | Output        |
| ------------- | ------------- |
| Hours Studied | Marks         |
| Area          | House Price   |
| Symptoms      | Disease       |
| Email         | Spam/Not Spam |


Why is it called supervised?

It is called Supervised Learning because the model learns under the supervision of correct answers (labels).

Think of it like a student learning with a teacher:

Teacher → Provides questions and correct answers.

Student → Learns the pattern from those examples.

Exam → Student predicts answers for new questions.

In Machine Learning:

Features (X) → Questions/Input

Labels (y) → Correct Answers/Output

Model → Student

**Example 1: House Price Prediction**

| Area (sq ft) | Bedrooms | Price   |
| ------------ | -------- | ------- |
| 1000         | 2        | 40 Lakh |
| 1500         | 3        | 60 Lakh |
| 2000         | 4        | 85 Lakh |

The model sees both:

Inputs: Area, Bedrooms

Correct output: Price

It learns the relationship and can then predict the price of a new house.

**Example 2: Spam Detection**

| Email Text          | Label    |
| ------------------- | -------- |
| "Win a free iPhone" | Spam     |
| "Meeting at 3 PM"   | Not Spam |

The model learns from emails that already have the correct labels.

Why "Supervised"?

Because during training, the algorithm is guided (supervised) by the correct output values.

In contrast, in Unsupervised Learning, there are no correct answers (labels).


    Input (X) + Correct Output (y)

            ↓

      Model learns patterns

            ↓

    Predicts output for new data

Final Understanding based on above examples:

Supervised Learning is a type of machine learning in which the model is trained using labeled data, where both the input features and the correct output are provided, allowing the model to learn the mapping from inputs to outputs.

**Types of Supervised Learning:**

1. Regression

Regression is a machine learning technique used to predict a continuous numeric value based on input features.

Example:

| Input Features        | Predicted Output                  |
| --------------------- | --------------------------------- |
| Area, Bedrooms        | House Price (₹50,00,000)          |
| Experience, Skills    | Salary (₹8,50,000)                |
| Advertising Budget    | Sales Amount (₹1,20,000)          |
| Temperature, Humidity | Electricity Consumption (250 kWh) |


2. Classification

Classification is a type of supervised learning in which the target variable (output) is a category or label.

| Input Features       | Predicted Output     |
| -------------------- | -------------------- |
| Email Text           | Spam / Not Spam      |
| Student Marks        | Pass / Fail          |
| Medical Data         | Disease / No Disease |
| Customer Information | Will Buy / Won't Buy |


Types of Classification:

| Type                       | Number of Classes            | Example         |
| -------------------------- | ---------------------------- | --------------- |
| Binary Classification      | 2                            | Spam / Not Spam |
| Multi-class Classification | More than 2 (one class only) | A, B, C, D      |
| Multi-label Classification | Multiple labels per sample   | Action + Comedy |
