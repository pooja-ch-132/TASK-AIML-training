3.Adding columns:

Create a new column called Grade using the following criteria.

```python
def assign_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 80:
        return "B"
    elif mark >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["marks"].apply(assign_grade)

print("DataFrame with Grade Column:")
print(df)
```

```python
#4. Modifying Columns
#Convert names to uppercase
df["student_name"] = df["student_name"].str.upper()

print("Names converted to uppercase:")
print(df)
```

```python
# 5. Removing Columns
# Remove Department column
df_without_dept = df.drop("department", axis=1)

print("DataFrame after removing Department:")
print(df_without_dept)
```

```python
# 6. Removing Rows
# Remove student record at index 3
df_removed_row = df.drop(index=3)
```

```python
# 7. GroupBy
# Average marks by department
print("Average Marks by Department:")
avg_marks = df.groupby("department")["marks"].mean()
print(avg_marks)
```

```python
# 8. Aggregate Functions
print("Aggregate Statistics for Marks:")

print("Mean:", df["marks"].mean())
print("Maximum:", df["marks"].max())
print("Minimum:", df["marks"].min())
print("Sum:", df["marks"].sum())
print("Median:", df["marks"].median())
print("Standard Deviation:", df["marks"].std())
```

```python
# 9. Combining DataFrames
new_data = {
    "student_name": ["Teja", "Divya"],
    "age": [20, 21],
    "marks": [87, 91],
    "department": ["CSE", "ECE"]
}

new_df = pd.DataFrame(new_data)
# Match existing columns
new_df["exam_date"] = pd.NaT
new_df["grade"] = new_df["marks"].apply(assign_grade)

combined_df = pd.concat([df, new_df], ignore_index=True)

print("Combined DataFrame:")
print(combined_df)
```

```python
# 10. Working with Dates
df["Exam_date"] = pd.to_datetime(df["exam_date"])

df["day"] = df["exam_Date"].dt.day
df["month"] = df["exam_Date"].dt.month
df["year"] = df["exam_Date"].dt.year

print("extracted day, month, year:")
print(df[["exam_date", "day", "month", "year"]])
```

```python
# 11. Indexing
indexed_df = df.set_index("Name")

print("Details of student POOJA MENON:")
print(indexed_df.loc["POOJA MENON"])
```

```python
# 12. Pivot Table
# Average Marks for each Department
pivot = pd.pivot_table(
    df,
    values="marks",
    index="department",
    aggfunc="mean"
)

print("Pivot Table (Average Marks by Department):")
print(pivot)
```
