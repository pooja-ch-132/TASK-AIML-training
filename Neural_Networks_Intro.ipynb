**Introduction to Neural Networks**

A Neural Network is a machine learning model inspired by the structure and functioning of the human brain. It consists of interconnected processing units called neurons, which work together to recognize patterns, classify data, and make predictions.


**What is a Perceptron?**

A Perceptron is the simplest type of artificial neural network and serves as the basic building block of more complex neural networks. It is mainly used for binary classification problems (e.g., Yes/No, True/False).

**1. Activation Functions**

An Activation Function decides whether a neuron should be activated (produce an output). It introduces non-linearity, allowing neural networks to solve complex problems beyond simple linear classification.

Without activation functions, a neural network behaves like a linear model regardless of its depth.

Common Activation Functions:

A. Step (Binary) Activation Function

Used in the original perceptron.

Formula:

      0,x<0​

Advantages:

1. Simple and easy to implement.
2. Suitable for binary classification.

Disadvantages:

1. Not differentiable.
2. Cannot be used effectively in deep learning.

**2. Sigmoid Function**

Formula: f(x) = 1/1+e-x

Output Range: 0 to 1

Advantages:

1. Smooth curve.
2. Useful for binary classification probabilities.

Disadvantages:

3. Suffers from the vanishing gradient problem.
4. Computationally expensive compared to ReLU.

Tanh Activation Function:

Definition

The Tanh (Hyperbolic Tangent) activation function is a nonlinear activation function used in neural networks. It converts the input value into an output between -1 and +1.

Unlike the sigmoid function, which outputs values between 0 and 1, tanh is zero-centered, making it easier for neural networks to learn.

where:

e = Euler's number (≈ 2.718)
x = Input to the neuron

Advantages
1. Zero-centered output
Produces both positive and negative outputs.
Helps gradients move in both directions.
Usually converges faster than Sigmoid.
2. Non-linear
Can learn complex patterns.
Suitable for deep neural networks.
3. Smooth gradient
Continuous and differentiable everywhere.
Supports backpropagation effectively.
4. Better than Sigmoid
Since outputs are centered around zero, optimization is often more efficient.

Disadvantages:

1. Vanishing Gradient Problem

For very large positive or negative inputs, the gradient becomes nearly zero.

Example:

tanh(10) ≈ 1
tanh(-10) ≈ -1

During backpropagation:

Gradient ≈ 0
Weights update very slowly
Deep networks become difficult to train

2. Computationally expensive: Uses exponential calculations (ex and e−x), making it slower than ReLU.

3. Saturation: For large input values, the output saturates at -1 or +1, reducing learning speed.



**ReLu**: ReLU (Rectified Linear Unit) Activation Function

The ReLU (Rectified Linear Unit) activation function is one of the most widely used activation functions in deep learning. It outputs the input directly if it is positive; otherwise, it outputs 0.

ReLU introduces non-linearity while being computationally simple, making it the default choice for most hidden layers in modern neural networks.

Formula: ReLU(x)=max(0,x)​

where: x = Input to the neuron

Advantages

1. Simple and Fast
Requires only a comparison with zero.
Computationally much faster than Sigmoid and Tanh.
2. Reduces Vanishing Gradient Problem
For positive inputs, the gradient is 1, allowing gradients to flow through deep networks.
Enables efficient training of very deep neural networks.
3. Sparse Activation
Negative inputs become zero.
Only some neurons are active, making the network more efficient.
4. Faster Training
Converges faster than Sigmoid and Tanh because of its simple computation and non-saturating positive region.

Disadvantages:

1. Dying ReLU Problem

If a neuron's inputs are always negative:

Output remains 0
Gradient becomes 0
The neuron stops learning permanently.

This is known as the Dead (Dying) ReLU problem.

2. Not Zero-centered
Outputs are always non-negative.
This can sometimes slow optimization compared to zero-centered activations.

3. Unbounded Positive Output
Very large positive inputs produce very large outputs.
In some cases, this can lead to unstable training if not managed properly.
