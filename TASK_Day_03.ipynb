```python
#Q1. Check whether a number is an Armstrong number
#An Armstrong number is a number that is equal to the sum of its digits,
#where each digit is raised to the power of the total number of digits. Ex:153=1+125+127=153
num = int(input("Enter a number: "))
temp = num
order = len(str(num))
sum_val = 0

while temp > 0:
    digit = temp % 10
    sum_val += digit ** order
    temp //= 10

if sum_val == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
#Q2. Find the greatest ncommon divisor(GCD) of two numbers
#Find the Greatest Common Divisor (GCD) of Two Numbers
#GCD(a, b) = GCD(b, a % b)
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    a, b = b, a % b

print("GCD =", a)
```

```python
#Q3. Find the least common multiple(LCM) of to numbers
#The LCM of two numbers is the smallest positive number that is divisible by both numbers.
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

greater = max(a, b)

while True:
    if greater % a == 0 and greater % b == 0:
        lcm = greater
        break
    greater += 1

print("LCM =", lcm)
```

```python
#Q4. Print all prime numbers between 1 & 100.
#Print All Prime Numbers Between 1 and 100
print("Prime numbers between 1 and 100:")

for num in range(2, 101):
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, end=" ")
```

```python
#Q5. Count vowels and consonants in a string.
#Count Vowels and Consonants in a String, Vowels = open sound (A, E, I, O, U)
#Consonants = all the rest of the letters
text = input("Enter a string: ")

vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0

for ch in text:
    if ch.isalpha():
        if ch in vowels:
            vowel_count += 1
        else:
            consonant_count += 1

print("Vowels:", vowel_count)
print("Consonants:", consonant_count)
```

```python
#Q6. Find the largest element in a list without using max()
#Find the Largest Element in a List Without Using max()
numbers = [12, 45, 67, 23, 89, 34]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest element:", largest)
```

```python
#Q7. Remove duplicate elements from a list.
#Remove Duplicate Elements from a List
numbers = [1, 2, 3, 2, 4, 5, 1, 6]

unique_list = []

for num in numbers:
    if num not in unique_list:
        unique_list.append(num)

print("Original List:", numbers)
print("List without duplicates:", unique_list)
```

```python
#Q8. Create a number guessing game using loops & conditions.
#Number Guessing Game
import random

secret_number = random.randint(1, 10)
while True:
    guess = int(input("Guess a number (1-10): "))

    if guess < secret_number:
        print("Too low!")
    elif guess > secret_number:
        print("Too high!")
    else:
        print("Congratulations! You guessed correctly.")
        break
```

```python
#Q9. Simulate and ATM menu using while and if-elif-else.

#Simulate an ATM Menu Using while and if-elif-else
balance = 10000

while True:
    print("\nATM MENU")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("Balance:", balance)

    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Updated Balance:", balance)
    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))

        if amount <= balance:
            balance -= amount
            print("Updated Balance:", balance)
        else:
            print("Insufficient Balance")

    elif choice == 4:
        print("Thank you for using ATM")
        break

    else:
        print("Invalid Choice")
```

```python
#Q10. Generate patterns using nested loops.
#Generate Patterns Using Nested Loops
rows = 5

for i in range(1, rows + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
#Q11. Print Pascals Triangle.
#Print Pascal's Triangle
rows = int(input("Enter number of rows: "))

for i in range(rows):
    num = 1

    for j in range(rows - i - 1):
        print(" ", end="")

    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)

    print()
```

```python
#Q12. Create a menu driven program that repeatedly asks the user to choose an operation until they select exit.
#Menu driven program
while True:
    print("\nMENU")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = int(input("Enter your choice: "))
    if choice == 5:
        print("Program Exited")
        break

    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if choice == 1:
        print("Result =", a + b)

    elif choice == 2:
        print("Result =", a - b)

    elif choice == 3:
        print("Result =", a * b)
    elif choice == 4:
        if b != 0:
          print("Result =", a / b)
        else:
          print("Division by zero not allowed")
    else:
        print("Invalid Choice")
```
