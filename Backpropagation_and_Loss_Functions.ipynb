**Backpropagation and Loss Functions**

 Introduction:

Deep learning models learn by reducing errors. The learning process consists of two main concepts:

Loss Function → Measures how wrong the prediction is.

Backpropagation → Updates the neural network's weights to reduce the loss.

Think of a neural network as a student preparing for an exam:

Makes an answer (prediction)
Compares with the correct answer
Calculates mistakes (loss)
Learns from mistakes (backpropagation)
Improves for the next attempt

This cycle repeats thousands of times until the network becomes accurate.

**What is a Loss Function?**

A loss function measures the difference between:

Actual output (Ground Truth)
Predicted output

It tells us:

"How bad is the model's prediction?"

Formula
Loss=f(Actual,Predicted)

If prediction is perfect

Loss = 0

If prediction is poor

Loss becomes larger.

Example

Suppose we predict house prices.

| Actual  | Predicted |
| ------- | --------- |
| 50 lakh | 48 lakh   |
| 75 lakh | 80 lakh   |
| 60 lakh | 62 lakh   |

Loss tells us how far predictions are from actual values.

**Why Do We Need a Loss Function?**

Without a loss function:

The network does not know whether it is correct.
It has no direction for improvement.

The loss acts like a "score" that guides learning.

**What is Backpropagation?**

Backpropagation is the learning algorithm of neural networks.

It sends the error backward through the network to adjust the weights.

Simple definition:

Backpropagation computes how much each weight contributed to the error and changes it to reduce future error.

Suppose prediction

95

Actual

100

Error

5

**********Which neuron caused this error?**********

Backpropagation answers this question.

It identifies which weights need adjustment.

Example:

**Relationship Between Loss Function and Backpropagation**

| Loss Function                 | Backpropagation                  |
| ----------------------------- | -------------------------------- |
| Measures prediction error     | Computes gradients of the error  |
| Produces a single error value | Uses the error to update weights |
| Indicates model performance   | Helps the model learn            |


**Summary**:

1. Loss function: Measures how wrong the prediction is.
2. Backpropagation: Calculates how to change the weights to reduce that error.
3. Gradient descent: Updates the weights using the gradients computed by backpropagation.
