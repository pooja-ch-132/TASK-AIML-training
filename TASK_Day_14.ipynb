*What is Plotly?*

Plotly is a data visualization library that enables users to create interactive, publication ready charts and dashboards in Python, R and JavaScript. It is widely used for exploratory data analysis, business reporting and web‑based visualisations.

We can create interactive visualizations such as:

Line charts,
Bar charts,
Scatter plots,
Maps,
Heatmaps,
3D plots,
Interactive dashboards.

Plotly is often combined with Dash to create full web dashboards.

Plotly consists of two key modules:

plotly.graph_objects: Low-level API of Plotly that contains objects such as Figure, layout and data which are responsible for plotting.

plotly.express: This is a high-level interface that simplifies the process of creating complex visualizations and automatic styling.

```python
import plotly.express as px

# Built-in dataset
df = px.data.gapminder()

# Filter for one year
df = df[df["year"] == 2007]

# Create interactive chart
fig = px.scatter(
    df,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    log_x=True,
    title="World Development Indicators (2007)",
    labels={
        "gdpPercap": "GDP Per Capita",
        "lifeExp": "Life Expectancy"
    }
)

fig.show()
```

```python
#Employee Count by Department Using bar plot
import pandas as pd
import plotly.express as px

# Sample data
df = pd.DataFrame({
    'EmpName': ['Ravi', 'Vikas', 'Kishor', 'Divya', 'Evan', 'Fiza', 'Gayatri'],
    'Department': ['HR', 'IT', 'IT', 'Finance', 'HR', 'Sales', 'IT']
})

# Count employees in each department
dept_count = df['Department'].value_counts().reset_index()
dept_count.columns = ['Department', 'Employee Count']

# Create bar chart
fig = px.bar(
    dept_count,
    x='Department',
    y='Employee Count',
    color='Department',
    title='Employee Count by Department'
)

fig.show()
```
