```python
#List -- A list in Python is an ordered and mutable collection of items.
# Elements can be added, removed, can be accessed by index value
num=[1,2,3,4,5,2,3.0,"heelo"]
print(num)
```

**When to use list**

1.Order matters

Example: Steps in a recipe, ranking of students, daily schedule.

["Wake up", "Exercise", "Work"]

2.Duplicates are allowed or meaningful

Example: Shopping cart items.

["apple", "banana", "apple"]

3.You need to access items by position (index)

Example: Get the first, second, or last item.

students[0]

4.You want to preserve the order in which items were added

Example: Playlist songs.

```python
#Set -- A set is an unordered collection of unique elements. It automatically removes duplicate values.
setnum={1,2,2,3,4,5,6,6,8,9,4}
print(setnum)
```

**Set Operations**

Union (|) Combines all unique elements.

Intersection (&) Finds common elements.

Difference (-) Elements in the first set but not in the second.

```python
#Union
a={1,2,3,3}
b={4,5,6,3}
print(a|b)
```

```python
#Interset
a={1,2,3}
b={4,5,3}
print(a&b)
```

```python
#difference
a={1,2,3}
b={4,5,3}
print(a-b)
```

**When to Use a Set**

Use a set when you need:

Unique values only

Fast membership checks (in)

Removing duplicates

Mathematical set operations (union, intersection, difference)

Mini Project:

Create a Student Report Card Program using:

1.List → store student names

2.Set → store subjects

3.Tuple → store student information (name, age)

4.Dictionary → store subject-wise marks

Requirements:

1.Ask the user to enter:

*Student name

*Age

*Number of subjects

2.For each subject:

*Enter subject name

*Enter marks

3.Store:

*Name in a list

*Subjects in a set

*Student info in a tuple

*Marks in a dictionary

4.Print a report card showing:

*Name

*Age

*Subjects

*Marks

*Total marks

*Average marks

```python
name = input("Enter student name: ")
age = int(input("Enter age: "))

students = []
subjects = set()
marks = {}

students.append(name)

student_info = (name, age)

n = int(input("How many subjects? "))

for i in range(n):
    sub = input("Enter subject name: ")
    mark = int(input("Enter marks: "))

    subjects.add(sub)
    marks[sub] = mark

total = sum(marks.values())
average = total / len(marks)

print("\n----- REPORT CARD -----")

print("Name:", student_info[0])
print("Age:", student_info[1])

print("\nSubjects:")
for subject in subjects:
    print(subject)

print("\nMarks:")
for subject, mark in marks.items():
    print(subject, ":", mark)

print("\nTotal Marks:", total)
print("Average Marks:", average)
```
