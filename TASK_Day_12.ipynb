# Seaborn
What is Seaborn?

Seaborn is a high-level data visualization library built on top of Matplotlib. It provides an easy way to create attractive and informative statistical graphics with less code.

*Built on top of Matplotlib and integrated with pandas data structures, Seaborn makes data visualization easier and more consistent.

*Seaborn emphasizes visualization as an essential part of data analysis.

*Its dataset-oriented APIs allow switching between different plot types for the same variables.

*Helps in understanding patterns, trends and relationships within the data.


Categories of Plots in Seaborn

1. Relational plots: used to understand the relation between two variables (e.g., lineplot, scatterplot)

2. Categorical plots: deals with categorical variables and how they can be visualized (e.g., boxplot, violinplot, barplot)

3. Distribution plots: show data distribution (e.g., histplot, kdeplot)

4. Regression plots: show relationships with regression line (e.g., lmplot, regplot)
Matrix plots: visualize matrices (e.g., heatmap)
Multi-plot grids: plot multiple graphs together (e.g., FacetGrid)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

students = pd.DataFrame({
    'Student':['Ravi','Madhu','Sam','Rosy','Neha','Anu','Babu','Swetha'],
    'Gender':['Male','Female','Male','Female',
              'Male','Female','Male','Female'],
    'Hours_Studied':[2,5,1,6,4,7,3,8],
    'Marks':[60,85,50,90,75,95,65,98]
})
```

```python
#Line plot:

import matplotlib.pyplot as plt
import seaborn as sns

#List
students = ["Ravi","Madhu","Sam","Rosy","Neha","Anu","Babu","Swetha"]
marks = [60,85,50,90,75,95,65,98]

# Matplotlib
plt.style.use('default')
plt.figure(figsize=(12,5))
plt.plot(students, marks)
plt.title("Matplotlib")
plt.show()

# Seaborn
sns.set_style("darkgrid")
plt.figure(figsize=(12,5))
sns.lineplot(x=students, y=marks)
plt.title("Seaborn")
plt.show()
```

```python
#2. Bar chart

plt.figure(figsize=(12,5))

avg_marks = students.groupby('Gender')['Marks'].mean()

# Matplotlib
plt.style.use('default')
plt.subplot(1,2,1)
plt.bar(avg_marks.index,
        avg_marks.values)
plt.title("Matplotlib Bar Plot")

# Seaborn
sns.set_style("darkgrid")
plt.subplot(1,2,2)
sns.barplot(data=students,
            x='Gender',
            y='Marks')
plt.title("Seaborn Bar Plot")

plt.tight_layout()
plt.show()
```

```python
#3. Histogram

plt.figure(figsize=(12,5))

# Matplotlib
plt.style.use('default')
plt.subplot(1,2,1)
plt.hist(students['Marks'],
         bins=5)
plt.title("Matplotlib Histogram")

# Seaborn
sns.set_style("dark")
plt.subplot(1,2,2)
sns.histplot(data=students,
             x='Marks',
             bins=5)
plt.title("Seaborn Histogram")

plt.tight_layout()
plt.show()
```

```python
#Line plot Excercise

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
flights = sns.load_dataset('flights')

# Create datetime column
flights['date'] = pd.to_datetime(
    flights['year'].astype(str) + '-' +
    flights['month'].astype(str)
)

# Plot
plt.figure(figsize=(12, 6))

plt.plot(
    flights['date'],
    flights['passengers'],
    color='blue',
    linestyle='--',
    linewidth=2,
    label='Passengers'
)

# Find peak
peak_idx = flights['passengers'].idxmax()
peak_date = flights.loc[peak_idx, 'date']
peak_value = flights.loc[peak_idx, 'passengers']

# Annotate peak
plt.annotate(
    f'Peak = {peak_value}',
    xy=(peak_date, peak_value),
    xytext=(peak_date, peak_value + 50),
    arrowprops=dict(arrowstyle='->')
)

# Formatting
plt.title('Monthly Airline Passengers Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Passengers')
#A legend is an area describing the elements of the graph
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
