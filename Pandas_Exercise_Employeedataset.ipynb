```python
import pandas as pd
df= pd.read_csv("employee_data.csv")
```

```python
print("Original Dataset:")
print(df)
```

```python
#Find Missing Values in Each Column
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Replace Missing Names with "Unknown"
df["Name"] = df["Name"].fillna("Unknown")
```

```python
# Fill Missing Departments with Most Frequent Department
most_freq_dept = df["Department"].mode()[0]
df["Department"] = df["Department"].fillna(most_freq_dept)
```

```python
# Replace Missing Salaries with Median Salary
median_salary = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(median_salary)
```

```python
# Replace Missing Experience with Average Experience
avg_exp = df["Years_of_Experience"].mean()
df["Years_of_Experience"] = df["Years_of_Experience"].fillna(avg_exp)
```

```python
# Rename Columns to Lowercase
df.columns = df.columns.str.lower()
```

```python
print("Dataset After Handling Missing Values:")
print(df)
```

```python
# Remove Duplicates
df = df.drop_duplicates()
```

```python
print("Dataset After Removing Duplicates:")
print(df)
```

```python
# Display All Employees from IT Department
print("Employees in IT Department:")
print(df[df["department"] == "IT"])
```

```python
# Employees Whose Salary > 70000
print("Employees with Salary > 70000:")
print(df[df["salary"] > 70000])
```

```python
# Employees Working in Hyderabad
print("Employees in Hyderabad:")
print(df[df["city"] == "Hyderabad"])
```

```python
# Employees with More Than 5 Years Experience
print("Employees with Experience > 5 Years:")
print(df[df["years_of_experience"] > 5])
```

```python
# Employees in IT or Finance Department
print("Employees in IT or Finance:")
print(df[df["department"].isin(["IT", "Finance"])])
```

```python
# Employees with Salary Between 60000 and 80000
print("Employees with Salary Between 60000 and 80000:")
print(df[df["salary"].between(60000, 80000)])
```

```python
# Number of Employees in Each Department
print("Employee Count by Department:")
print(df["department"].value_counts())
```

```python
# Total Salary Paid by Each Department
print("Total Salary by Department:")
print(df.groupby("department")["salary"].sum())
```

```python
# Average Salary by City
print("Average Salary by City:")
print(df.groupby("city")["salary"].mean())
```

```python
# Multiple Aggregations Department-wise
print("Department-wise Salary Statistics:")

dept_stats = df.groupby("department").agg(
    Average_Salary=("salary", "mean"),
    Maximum_Salary=("salary", "max"),
    Minimum_Salary=("salary", "min"),
    Total_Salary=("salary", "sum")
)

print(dept_stats)
```

```python
# Pivot Table - Average Salary by Department
print("Pivot Table: Average Salary by Department")

pivot_table = pd.pivot_table(
    df,
    values="salary",
    index="department",
    aggfunc="mean"
)

print(pivot_table)
```
