What is Linear Regression?

Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value.

Examples:

Predicting house prices

Predicting sales revenue

Predicting a student's marks

Predicting temperature

It finds the best-fit straight line that describes the relationship between input variables (features) and the output variable (target).



**Linear Regression Equation:**

The equation of a simple linear regression line is:

y=mx+b

where:

y = predicted value

x = input feature

m = slope (how much y changes when x changes)

b = intercept (value of y when x = 0)




Example Problem

Suppose we want to predict a student's exam marks based on the number of hours studied.

| Hours Studied (x) | Marks (y) |
| ----------------- | --------- |
| 1                 | 45        |
| 2                 | 50        |
| 3                 | 55        |
| 4                 | 60        |
| 5                 | 65        |
| 6                 | 70        |

The relationship is approximately:

Marks=5×Hours+40

If a student studies for 7 hours:

Marks=5(7)+40=75


Python Example using Scikit-Learn

Scikit usually refers to a family of scientific and machine learning libraries built on top of packages like NumPy and SciPy. The most popular one is:

scikit-learn (often imported as sklearn)

scikit-learn is an open-source library used for:

Machine learning algorithms

Classification (e.g., spam detection)
Regression (e.g., predicting house prices)
Clustering (e.g., customer segmentation)

Data preprocessing

Scaling features
Handling missing values
Encoding categorical data

Model evaluation and selection

Train-test splitting
Cross-validation
Performance metrics

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Input data (Hours studied)
X = np.array([[1],
              [2],
              [3],
              [4],
              [5],
              [6]])

# Output data (Marks)
y = np.array([45, 50, 55, 60, 65, 70])

# Create model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict marks for 7 hours of study
prediction = model.predict([[7]])

print("Predicted Marks:", prediction[0])

print("Slope (m):", model.coef_[0])
print("Intercept (b):", model.intercept_)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([45, 50, 55, 60, 65, 70])

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

print("Actual:", y_test)
print("Predicted:", y_pred)
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```

Advantages of Linear Regression

✅ Simple and easy to understand
✅ Fast training and prediction
✅ Easy to interpret coefficients
✅ Works well when the relationship is approximately linear

Disadvantages

❌ Assumes a linear relationship between variables
❌ Sensitive to outliers
❌ Does not perform well on complex nonlinear data
❌ Can suffer from multicollinearity
